# 1. Download player stats

This notebook downloads regular-season, per-game stats for 2020–21 through 2024–25 and saves one CSV per season under `data/raw/statistics/`. Run it from the repository root. It uses the NBA Stats API through the `nba_api` package.

```python
from pathlib import Path
import time

import pandas as pd
from nba_api.stats.endpoints import leaguedashplayerstats

ROOT = Path.cwd()
DATA_DIR = ROOT / "data" / "raw" / "statistics"
DATA_DIR.mkdir(parents=True, exist_ok=True)

SEASONS = ["2020-21", "2021-22", "2022-23", "2023-24", "2024-25"]
```

## Download each season

The API request can take a few seconds. The short pause between requests helps avoid sending them back-to-back.

```python
for season in SEASONS:
    stats = leaguedashplayerstats.LeagueDashPlayerStats(
        season=season,
        per_mode_detailed="PerGame",
    ).get_data_frames()[0]

    filename = f"player_stats_{season[:4]}.csv"
    path = DATA_DIR / filename
    stats.to_csv(path, index=False)
    print(f"{season}: saved {len(stats):,} player rows to {path.relative_to(ROOT)}")
    time.sleep(1)
```

## Check the downloaded files
